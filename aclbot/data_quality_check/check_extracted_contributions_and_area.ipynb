```python
import json
from pathlib import Path
from prompts.schemas import ContributionType, Area
```

## Load data

```python
data_dir = Path('../data/datasets/acl-anthology')
filename = 'contributions_and_area.json'
with open(data_dir / filename) as f:
    data = json.load(f)
```

## Basic data analysis
Count the number of entries


```python
print(len(data))
```

### Functions

```python
def check_for_allowed_values(
    values: list,
    allowed_values: list
):
    n_not_allowed = 0
    for v in values:
        if v not in allowed_values:
            n_not_allowed += 1
    print(f'Found {n_not_allowed} not allowed values out of {len(values)} in total.')
```

### Contributions

Check that all contributions are in the list of allowed contributions

```python
contributions_choices = [
    entry['choice']
    for paper_id, extracted_data_for_paper in data.items()
    for entry in extracted_data_for_paper['contributions']
]
allowed_contribution_types = [contribution_type.value for contribution_type in ContributionType]
check_for_allowed_values(contributions_choices, allowed_contribution_types)
```

### Area

Check that all areas are in list of allowed areas

```python
area_choices = [
    extracted_data_for_paper['area']['choice']
    for _, extracted_data_for_paper in data.items()
]
allowed_areas = [area.value for area in Area]
check_for_allowed_values(area_choices, allowed_areas)
```
